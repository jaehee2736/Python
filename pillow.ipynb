```python
from PIL import Image
```

```python
img = Image.open("pngwing.com.png")

img.show()
```

```python
size_img = img.resize((300,300))
size_img.show()

print(img.size)
print(size_img.size)
```

```python
crop_img = img.crop((100,100,300,300))
crop_img.show()

crop_img.save("crop_wing.png")
```

```python
rotate_img = img.rotate(90)

rotate_img.show()
```

```python
color_img = img.convert("L")

color_img.show()
```

```python
img = Image.open("pngwing.com.png")

img = img.convert("L")
img = img.crop((400, 0 , 800, 1024))
img = img.rotate(20,expand=True)
img = img.resize((500,500))

img.show()
```

```python
img = Image.open("pngwing.com.png")
img = img.resize((500,500))
img = img.rotate(45, expand=True)
img = img.convert("L")
img.save("wing_png_A.png")
img.show()
```
