```python
ad = {"name": ["lee","kim","park","choi"],
      "score": [90,78,65,85]
      }

df = pd.DataFrame(ad)
print(df)

print(len(df))
print(df["score"].mean())
print(df["score"].max()) 
print(df["score"].min()) 
```

```python
bb = ["가위","바위","보"]
me = str(input("가위, 바위, 보 中"))
if me == "가위" or "바위" or "보" :
    enemy = random.choice(bb)
    if me == enemy :
        print()
    
else :
    print("기권패 수고")

```
